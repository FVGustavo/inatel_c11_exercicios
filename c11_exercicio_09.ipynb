```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Questão 1

```python
iris = sns.load_dataset("iris")
corr = iris[iris["species"] == "setosa"].corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.show()
```

Questão 2

```python
titanic = sns.load_dataset("titanic")

sns.histplot(data=titanic, x="age", hue="sex", kde=True)

plt.show()
```

Questão 3

```python
titanic = sns.load_dataset("titanic")

sns.boxplot(data=titanic, x="class", y="age", hue="sex")

plt.show()
```

Questão 4

```python
mpg = sns.load_dataset('mpg')

sns.regplot(data=mpg, x="horsepower", y="mpg", color='red')

plt.show()
```

Questão 5

```python
paises = pd.read_csv("paises.csv", sep=";")

cols = ["GDP ($ per capita)", "Literacy (%)", "Infant mortality (per 1000 births)", "Phones (per 1000)"]
corr = paises[cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.show()
```
