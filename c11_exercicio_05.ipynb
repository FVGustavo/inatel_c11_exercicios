Preparação

```python
import numpy as np

dados = np.genfromtxt('/content/sample_data/space.csv', delimiter=';', dtype=str,
                      skip_header=1, encoding='utf-8-sig', autostrip=True)

empresa       = dados[:, 1]
local         = dados[:, 2]
status_rocket = dados[:, 5]
status_missao = dados[:, 7]

custo = np.char.strip(dados[:, 6])
custo = np.where(custo == '', '0', custo).astype(float)

total = dados.shape[0]
```

Questão 1

```python
sucessos = np.sum(status_missao == 'Success')
perc_sucesso = sucessos / total * 100

print(f'Missões bem-sucedidas: {sucessos} de {total}')
print(f'Porcentagem de sucesso: {perc_sucesso:.2f}%')
```

Questão 2

```python
custos_validos = custo[custo > 0]
media_custo = custos_validos.mean()

print(f'Missões com custo informado: {custos_validos.size}')
print(f'Custo médio: {media_custo:.2f} milhões de USD')
```

Questão 3


```python
missoes_eua = np.sum(np.char.find(local, 'USA') >= 0)

print(f'Missões realizadas nos EUA: {missoes_eua}')
```

Questão 4

```python
mask_spacex   = empresa == 'SpaceX'
custos_spacex = custo[mask_spacex]
idx           = np.argmax(custos_spacex)

missao = dados[mask_spacex][idx]

print('Missão mais cara da SpaceX:')
print(f'  Detalhe: {missao[4]}')
print(f'  Data...: {missao[3]}')
print(f'  Local..: {missao[2]}')
print(f'  Custo..: {custos_spacex[idx]:.2f} milhões de USD')
```

Questão 5

```python
nomes, quantidades = np.unique(empresa, return_counts=True)

ordem = np.argsort(quantidades)[::-1]
nomes, quantidades = nomes[ordem], quantidades[ordem]

for nome, qtd in zip(nomes, quantidades):
    print(f'{nome}: {qtd} missões')
```

Questão 6

```python
retirados = np.sum(status_rocket == 'StatusRetired')
perc_retirados = retirados / total * 100

print(f'Missões com foguetes aposentados: {retirados} de {total}')
print(f'Porcentagem: {perc_retirados:.2f}%')
```
