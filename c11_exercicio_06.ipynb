Questão 1

```python
import pandas as pd
import numpy as np

seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

print(seriesAno1)
print(seriesAno2)
```

Questão 2

```python
print(seriesAno1.sum())
print(seriesAno2.sum())
```

Questão 3


```python
crescimento = (seriesAno2 - seriesAno1) / seriesAno1 * 100

print(crescimento)
```

Questão 4

```python
print(crescimento[crescimento > 0])
```

Questão 5

```python
projecao = seriesAno2 * (1 + crescimento/100) ** 2
projecao.nlargest(1)
```

Questão 6

```python
df = pd.DataFrame(
    index = ['A', 'B', 'C', 'D', 'E'],
    columns = ['W', 'X', 'Y', 'Z'],
    data = np.random.randint(1, 50, [5, 4])
)
print(df)

menoresQue30 = df['X'][df['X'] < 30]
print(menoresQue30.mean())
```
